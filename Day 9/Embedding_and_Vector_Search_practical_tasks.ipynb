Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
# place your code here
def cos_similarity_calculation(text,list_sentences):
    text_embed = model.encode(text)
    list_embed = model.encode(list_sentences)
    results = {}
    for og,vector in zip(list_sentences,list_embed):
        cosine_similarity = util.cos_sim(text_embed,vector)
        results[og] = float(cosine_similarity)
    results_sorted = dict(sorted(results.items(), key=lambda item: item[1], reverse=True))
    return results_sorted

text = "Our planet's profound ocean depths represent a vast, mysterious frontier filled with hidden secrets lurking beneath heavy, dark waters."
scores = cos_similarity_calculation(text,sentences_list)
print(f"Input: {text}")
print(f"Output: {scores}")
print(f"Top matching outcome: {list(scores.keys())[0]}")

        

```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
# place your code here
import numpy as np

def cos_similarity_batch(vector,batch_vectors):
    vector_arr = np.array(vector)
    batch_vectors_arr = np.array(batch_vectors)
    dot_product = np.dot(batch_vectors_arr,vector_arr)
    vector_norm = np.linalg.norm(vector)
    batch_norm = np.linalg.norm(batch_vectors,axis=1)
    cosine_similarity = dot_product / (batch_norm * vector_norm)
    return cosine_similarity

vector = [1,2,3]
batch_vectors = [[1,2,3],[5,5,2],[-1,-3,-2],[3,0,-1]]

scores = cos_similarity_batch(vector,batch_vectors)
results = {}
for og,score in zip(batch_vectors,scores):
    results[str(og)] = float(score)
results_sorted = dict(sorted(results.items(), key=lambda item: item[1], reverse=True))
print(f"Input: {vector}")
print(f"Output: {results_sorted}")
print(f"Top matching outcome: {list(results_sorted.keys())[0]}")
```
